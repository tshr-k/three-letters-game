```python
import json
import pandas as pd
from pprint import pprint
from typing import Dict, List
```

```python
df = pd.read_csv("../data/words.csv", usecols=["Rank", "Word", "Frequency count"])
df.dropna(inplace=True)
df.rename(columns={"Frequency count": "Frequency"}, inplace=True)
df.set_index("Word", inplace=True)
df
```

```python
def contains_triplet(word: str) -> bool:
    for j in range(len(word) - 2):
        if len(set(word[j:j+3])) == 1:
            return True
    return False
```

```python
words = df.index

is_valid = [False] * len(words)
valid_word_idx = 0
trigram_to_word_indxs: Dict[str, List[int]] = dict()

for i, word in enumerate(words):
    length = len(word)
    if length < 3 or contains_triplet(word): continue
    for j in range(length-2):
        trigram = word[j:j+3]
        if trigram not in trigram_to_word_indxs: trigram_to_word_indxs[trigram] = []
        trigram_to_word_indxs[trigram].append(valid_word_idx)
    is_valid[i] = True
    valid_word_idx += 1
```

```python
valid_words = []
for i, word in enumerate(words):
    if is_valid[i]:
        valid_words.append(word) 
```

```python
with open("../data/trigram_to_word_idxs.json", "w") as f:
    json.dump(trigram_to_word_indxs, f)
pd.Series(valid_words).to_csv("../data/valid_words.csv", index=False, header=False)
```
